```python

# Identify repeated feature combinations
feature_columns = df.columns.drop("phishing")

feature_groups = df.groupby(
    list(feature_columns),
    dropna=False
)["phishing"].agg(["size", "nunique"])

# Count repeated groups and groups with conflicting labels
repeated_groups = feature_groups[feature_groups["size"] > 1]

conflicting_groups = repeated_groups[
    repeated_groups["nunique"] > 1
]

print("Repeated feature groups:", len(repeated_groups))
print("Groups with conflicting labels:", len(conflicting_groups))
print("\nExamples of conflicting groups:")
display(conflicting_groups.head(10))
```

```python

# Find the groups with conflicting phishing labels
conflicting_groups = feature_groups[
    (feature_groups["size"] > 1) &
    (feature_groups["nunique"] > 1)
]

print("Number of conflicting groups:", len(conflicting_groups))
print("Total rows in conflicting groups:", conflicting_groups["size"].sum())

print("\nConflicting group details:")
display(conflicting_groups)

# Inspect the original records in those groups
conflicting_index = df.set_index(
    list(feature_columns)
).index.isin(
    conflicting_groups.index
)

conflicting_rows = df.loc[conflicting_index]

print("\nOriginal records involved:")
display(conflicting_rows[["phishing"]].head(20))
```

```python

# Check -1 values across URL feature groups
feature_groups = {
    "URL": [c for c in df.columns if c.endswith("_url")],
    "Domain": [c for c in df.columns if c.endswith("_domain")],
    "Directory": [c for c in df.columns if c.endswith("_directory")],
    "File": [c for c in df.columns if c.endswith("_file")],
    "Parameters": [c for c in df.columns if c.endswith("_params")],
    "Other": [
        c for c in df.columns
        if not any(
            c.endswith(suffix)
            for suffix in [
                "_url", "_domain", "_directory", "_file", "_params"
            ]
        )
        and c != "phishing"
    ]
}

for group, columns in feature_groups.items():
    if columns:
        count = (df[columns] == -1).sum().sum()
        total = df[columns].size
        print(
            f"{group}: {count:,} values equal to -1 "
            f"out of {total:,} feature values "
            f"({count / total * 100:.2f}%)"
        )
```

```python

# Identify features that do not belong to the URL component groups
other_columns = [
    c for c in df.columns
    if not any(
        c.endswith(suffix)
        for suffix in [
            "_url", "_domain", "_directory", "_file", "_params"
        ]
    )
    and c != "phishing"
]

# Display how often each feature contains -1
other_summary = pd.DataFrame({
    "minus_one_count": (df[other_columns] == -1).sum(),
    "minus_one_percentage": (df[other_columns] == -1).mean() * 100
})

other_summary = other_summary.sort_values(
    "minus_one_percentage", ascending=False
)

display(other_summary.head(25))
```

```python

# Features with -1 values that need further investigation
features_to_check = [
    "time_domain_expiration",
    "time_domain_activation",
    "domain_spf",
    "qty_redirects",
    "asn_ip",
    "time_response",
    "ttl_hostname",
    "qty_ip_resolved",
    "url_google_index",
    "domain_google_index"
]

# Compare the percentage of -1 values within each target class
for feature in features_to_check:
    print(f"\n{feature}")
    
    summary = (
        df.assign(has_minus_one=df[feature].eq(-1))
        .groupby("phishing")["has_minus_one"]
        .agg(["sum", "count", "mean"])
    )
    
    summary["percentage"] = (summary["mean"] * 100).round(2)
    display(summary[["sum", "count", "percentage"]])
```

```python

# Compare -1 values by target label without modifying df
for feature in features_to_check:
    print(f"\n{feature}")

    has_minus_one = df[feature].eq(-1)

    summary = pd.DataFrame({
        "minus_one_count": has_minus_one.groupby(df["phishing"]).sum(),
        "total_records": has_minus_one.groupby(df["phishing"]).size(),
        "percentage": (
            has_minus_one.groupby(df["phishing"]).mean() * 100
        ).round(2)
    })

    display(summary)
```

```python

# Identify identical feature sets that have conflicting target labels
feature_columns = df.columns.drop("phishing")

feature_groups = (
    df.groupby(list(feature_columns), dropna=False)["phishing"]
    .agg(["size", "nunique"])
)

conflicting_groups = feature_groups[
    (feature_groups["size"] > 1) &
    (feature_groups["nunique"] > 1)
]

print("Conflicting feature groups:", len(conflicting_groups))
print("Rows in conflicting groups:", conflicting_groups["size"].sum())

# Display the records belonging to those groups
conflict_index = df.set_index(list(feature_columns)).index.isin(
    conflicting_groups.index
)

conflict_rows = df.loc[conflict_index].sort_values(
    list(feature_columns)
)

display(conflict_rows[["phishing"]].head(20))
```

```python

# Get the feature combinations that have conflicting labels
conflicting_features = conflicting_groups.index

# Match all original records belonging to those feature combinations
df_indexed = df.set_index(list(feature_columns))

conflict_mask = df_indexed.index.isin(conflicting_features)
conflict_records = df_indexed.loc[conflict_mask].reset_index()

# Display the feature values and labels for all conflicting records
display(
    conflict_records.sort_values(
        list(feature_columns) + ["phishing"]
    ).reset_index(drop=True)
)
```

```python

# Find conflicting feature combinations and their original row numbers
feature_columns = df.columns.drop("phishing")

conflicting_groups = (
    df.groupby(list(feature_columns), dropna=False)["phishing"]
    .agg(["size", "nunique"])
)

conflicting_groups = conflicting_groups[
    (conflicting_groups["size"] > 1) &
    (conflicting_groups["nunique"] > 1)
]

# Identify all rows that belong to the conflicting groups
indexed_features = pd.MultiIndex.from_frame(df[feature_columns])
conflict_mask = indexed_features.isin(conflicting_groups.index)

# Display a compact summary of the affected rows
conflict_summary = df.loc[conflict_mask, ["phishing"]].copy()
conflict_summary.insert(
    0, "original_row_number", conflict_summary.index
)

display(conflict_summary.sort_values("original_row_number"))
```

```python

# Preserve the original dataset and create a working copy
df_work = df.copy()

# Identify feature combinations with conflicting labels
feature_columns = df_work.columns.drop("phishing")

label_counts = (
    df_work.groupby(list(feature_columns), dropna=False)["phishing"]
    .nunique()
)

# Mark every row whose feature combination has conflicting labels
conflict_keys = label_counts[label_counts > 1].index

feature_index = pd.MultiIndex.from_frame(df_work[feature_columns])
conflict_mask = feature_index.isin(conflict_keys)

# Report the number of affected rows
print("Original dataset rows:", len(df_work))
print("Rows in conflicting groups:", int(conflict_mask.sum()))
print("Conflicting feature groups:", len(conflict_keys))
```

```python

# Identify the conflicting feature groups
feature_columns = df.columns.drop("phishing")

group_summary = (
    df.groupby(list(feature_columns), dropna=False)["phishing"]
    .agg(record_count="size", distinct_labels="nunique")
    .reset_index()
)

conflicts = group_summary[
    (group_summary["record_count"] > 1) &
    (group_summary["distinct_labels"] > 1)
].copy()

print("Number of conflicting feature groups:", len(conflicts))
print("Records across conflicting groups:", conflicts["record_count"].sum())

# Show each group's size and its conflicting labels
for group_number, (_, group) in enumerate(conflicts.iterrows(), start=1):
    features = group[feature_columns]
    mask = df[feature_columns].eq(features).all(axis=1)
    labels = sorted(df.loc[mask, "phishing"].unique().tolist())

    print(
        f"Group {group_number}: "
        f"{int(mask.sum())} records, labels = {labels}"
    )
```
